## The decision, action, and cost of a wrong call

**Research question:** Which pages should be prioritized for human review for refresh, expansion, protection, pruning, or monitoring based on observable search-performance, visibility, freshness, and engagement signals?

**Action:** The ranked pages would be reviewed by a human, who could decide whether a page should be refreshed, expanded, protected, pruned, or simply monitored.
